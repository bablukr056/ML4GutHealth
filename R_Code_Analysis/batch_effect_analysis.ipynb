```python
import pandas as pd
import numpy as np
from skbio.stats.distance import DistanceMatrix
from skbio.stats.ordination import pcoa
from scipy.spatial.distance import pdist, squareform
import matplotlib.pyplot as plt
import matplotlib as mpl
```

```python
meta_data = pd.read_csv("otu_data/metadata_for_7452.tsv", sep="\t")
training_otu_wihtoutcorrected = pd.read_csv("otu_data/without_corrected_otu_7452_species.csv", sep=",", index_col=0)
training_corrected = pd.read_csv("otu_data/corrected_training_7452.csv", sep="\t", header=0, index_col=0, engine="c")
validation_without_corrected = pd.read_csv("otu_data/without_corrected_validation_OTU_642_samples.csv", sep=",", index_col=0)
validation_corrected = pd.read_csv("otu_data/corrected_corrected_validation_OTU_642_samples.csv", sep=",", index_col=0)
validation_metadata = pd.read_csv("otu_data/validation_metadata_642_samples.csv", sep=",")

print(meta_data.shape)
print(training_otu_wihtoutcorrected.shape)
print(training_corrected.shape)
print(validation_without_corrected.shape)
print(validation_corrected.shape)
print(validation_metadata.shape)
```

```python
mpl.rcParams['pdf.fonttype'] = 42
mpl.rcParams['ps.fonttype'] = 42
mpl.rcParams['font.size'] = 11
mpl.rcParams['axes.titlesize'] = 12
mpl.rcParams['axes.labelsize'] = 11
```

```python
def compute_pcoa(otu_table):
    sample_by_species = otu_table.T
    bc_dist = pdist(sample_by_species.values, metric='braycurtis')
    bc_dist_matrix = squareform(bc_dist)
    dm = DistanceMatrix(bc_dist_matrix, ids=sample_by_species.index)
    pcoa_results = pcoa(dm)
    return pcoa_results, sample_by_species.index
```

```python
pcoa_train_before, train_ids = compute_pcoa(training_otu_wihtoutcorrected)
pcoa_train_after, _ = compute_pcoa(training_corrected)
pcoa_val_before, val_ids = compute_pcoa(validation_without_corrected)
pcoa_val_after, _ = compute_pcoa(validation_corrected)
```

```python
meta_data_indexed = meta_data.set_index("Run_ID")
validation_metadata_indexed = validation_metadata.set_index("Run_ID")

train_batch = meta_data_indexed.loc[train_ids, "BioProject"]
val_batch = validation_metadata_indexed.loc[val_ids, "BioProject"]

train_health = meta_data_indexed.loc[train_ids, "Health_status"]
val_health = validation_metadata_indexed.loc[val_ids, "Health_status"]
```

```python
def plot_pcoa(ax, pcoa_result, batch_labels, title):
    coords = pcoa_result.samples.iloc[:, [0, 1]]
    unique_batches = batch_labels.unique()
    cmap = plt.cm.get_cmap('tab20', len(unique_batches))
    
    for i, batch in enumerate(unique_batches):
        mask = (batch_labels == batch).values
        ax.scatter(coords.iloc[mask, 0], coords.iloc[mask, 1],
                   alpha=0.65, s=18, color=cmap(i), edgecolors='none')
    
    var_exp = pcoa_result.proportion_explained
    ax.set_xlabel(f"PCo1 ({var_exp.iloc[0]*100:.2f}%)")
    ax.set_ylabel(f"PCo2 ({var_exp.iloc[1]*100:.2f}%)")
    ax.set_title(title, fontweight='bold')
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 12), dpi=100)

plot_pcoa(axes[0,0], pcoa_train_before, train_batch,
          "Training Set — Before Batch Correction")
plot_pcoa(axes[0,1], pcoa_train_after, train_batch,
          "Training Set — After Batch Correction")
plot_pcoa(axes[1,0], pcoa_val_before, val_batch,
          "Validation Set — Before Batch Correction")
plot_pcoa(axes[1,1], pcoa_val_after, val_batch,
          "Validation Set — After Batch Correction")

plt.tight_layout()
```

```python
fig.savefig("PCoA_batch_correction_comparison.png", dpi=600, bbox_inches='tight')
fig.savefig("PCoA_batch_correction_comparison.pdf", bbox_inches='tight')
fig.savefig("PCoA_batch_correction_comparison.tiff", dpi=600, bbox_inches='tight')

plt.show()
```
